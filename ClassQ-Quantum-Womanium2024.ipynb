```python
from classiq import QArray, X, qfunc

@qfunc
def flip_msb(reg: QArray):
    X(reg[reg.len - 1])

print(X)
```

```python
from classiq import Output, QBit, QNum, allocate


@qfunc
def main(indicator: Output[QBit]):

    x = QNum("x")
    allocate(4, x)
    flip_msb(x)

    indicator |= x == 8
```

```python
from classiq import create_model, show, synthesize

quantum_program = synthesize(create_model(main))
show(quantum_program)
```

```python
from classiq import execute

job = execute(quantum_program)
results = job.result()[0].value.parsed_counts
print(results)
```

```python
from classiq import write_qmod

write_qmod(create_model(main), "quantum_variables_and_functions")
```

```python
from classiq import Output, QBit, QNum, allocate

@qfunc
def main(indicator: Output[QBit]):
    allocate(25, x);

    for i in range(4):
      hadamard_transform(x[i]);

    for i in range(4):
      y += x[i]**2+1;
                           
    indicator |= y
```

```python
@qfunc
def main(x: Output[QNum], y: Output[QNum]):

    allocate(4, x)
    hadamard_transform(x)  # creates a uniform superposition
    y |= x**2 + 1
```

```python
quantum_program = synthesize(create_model(main))
```

```python

quantum_model_with_constraints = set_constraints(
    quantum_model, Constraints(optimization_parameter="width", max_depth=500)
)

quantum_program = synthesize(quantum_model_with_constraints)
write_qmod(quantum_model_with_constraints, "optimize")
#job = execute(quantum_program)
```

```python
quantum_model_with_preferences = set_preferences(
    quantum_model,
    Preferences(backend_service_provider="IBM Quantum", backend_name="ibm_brisbane"),
)

quantum_program = synthesize(quantum_model_with_preferences)
```

```python
circuit_width = QuantumProgram.from_qprog(quantum_program).data.width
circuit_depth = QuantumProgram.from_qprog(quantum_program).transpiled_circuit.depth
print(f"The circuit width is {circuit_width} and the circuit_depth is {circuit_depth}")
```

```python
def mcx_balanced(control_qubits, target_qubit):
    # Use a combination of ancillary qubits and direct gates
    ancilla = allocate_ancilla(qubits=2)

    # Apply a sequence of Toffoli gates
    toffoli([control_qubits[0], control_qubits[1]], ancilla[0])
    toffoli([control_qubits[2], control_qubits[3]], ancilla[1])
    toffoli([ancilla[0], ancilla[1]], target_qubit)
    toffoli([control_qubits[2], control_qubits[3]], ancilla[1])  # Uncompute
    toffoli([control_qubits[0], control_qubits[1]], ancilla[0])  # Uncompute

    return
```

```python
from qiskit import QuantumCircuit
from qiskit.visualization import circuit_drawer
from qiskit.circuit.library.standard_gates.x import MCXGrayCode, MCXRecursive, MCXVChain
import os

def create_mcx_circuits():
    control_qubits = 5
    target_qubit = 1

    # Total number of qubits needed for each implementation
    total_qubits_depth = control_qubits + target_qubit + 1  # One ancillary qubit for depth-optimized
    total_qubits_width = control_qubits + target_qubit  # No ancillary qubits for width-optimized
    total_qubits_balanced = control_qubits + target_qubit + 2  # Two ancillary qubits for balanced approach

    # Minimized Depth Implementation
    qc_depth = QuantumCircuit(total_qubits_depth)
    mcx_depth = MCXGrayCode(control_qubits)
    qc_depth.append(mcx_depth, range(total_qubits_depth))

    # Minimized Width Implementation
    qc_width = QuantumCircuit(total_qubits_width)
    mcx_width = MCXVChain(control_qubits)
    qc_width.append(mcx_width, range(total_qubits_width))

    # Balanced Implementation
    qc_balanced = QuantumCircuit(total_qubits_balanced)
    mcx_balanced = MCXRecursive(control_qubits)
    qc_balanced.append(mcx_balanced, range(total_qubits_balanced))

    return qc_depth, qc_width, qc_balanced

# Create the circuits
circuits = create_mcx_circuits()

```
